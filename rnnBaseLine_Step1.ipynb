```python
# Comprehensive file structure check
import os

print("=" * 60)
print("CHECKING FILE STRUCTURE")
print("=" * 60)

# Check base directory
print(f"\n1. Base Directory: {baseDir}")
print(f"   Exists: {os.path.exists(baseDir)}")

if os.path.exists(baseDir):
    print(f"\n   Contents of base directory:")
    try:
        contents = os.listdir(baseDir)
        for item in sorted(contents):
            item_path = os.path.join(baseDir, item)
            item_type = "📁 DIR " if os.path.isdir(item_path) else "📄 FILE"
            print(f"   {item_type} - {item}")
    except Exception as e:
        print(f"   ❌ Error listing directory: {e}")
else:
    print(f"   ❌ BASE DIRECTORY NOT FOUND!")

# Check critical files
print(f"\n2. Required Python Files:")
required_files = [
    'makeTFRecordsFromSession.py',
    'getSpeechSessionBlocks.py'
]

for file in required_files:
    file_path = baseDir + '/' + file
    exists = os.path.exists(file_path)
    status = "✅ FOUND" if exists else "❌ MISSING"
    print(f"   {status} - {file}")

# Check neuralDecoder folder
print(f"\n3. neuralDecoder Folder:")
neural_decoder_path = baseDir + '/neuralDecoder'
exists = os.path.exists(neural_decoder_path)
status = "✅ FOUND" if exists else "❌ MISSING"
print(f"   {status} - neuralDecoder/")

if exists:
    print(f"   Contents of neuralDecoder/:")
    try:
        contents = os.listdir(neural_decoder_path)
        for item in sorted(contents):
            print(f"      - {item}")

        # Check for datasets subfolder
        datasets_path = neural_decoder_path + '/datasets'
        if os.path.exists(datasets_path):
            print(f"\n   ✅ neuralDecoder/datasets/ exists")
            speech_dataset_path = datasets_path + '/speechDataset.py'
            if os.path.exists(speech_dataset_path):
                print(f"   ✅ speechDataset.py exists")
            else:
                print(f"   ❌ speechDataset.py MISSING")
        else:
            print(f"   ❌ neuralDecoder/datasets/ folder MISSING")
    except Exception as e:
        print(f"   ❌ Error reading neuralDecoder: {e}")

# Check competitionData folder
print(f"\n4. competitionData Folder:")
competition_data_path = baseDir + '/competitionData'
exists = os.path.exists(competition_data_path)
status = "✅ FOUND" if exists else "❌ MISSING"
print(f"   {status} - competitionData/")

if exists:
    print(f"   Contents of competitionData/:")
    try:
        contents = os.listdir(competition_data_path)
        print(f"   Found {len(contents)} items:")
        for item in sorted(contents)[:10]:  # Show first 10 items
            print(f"      - {item}")
        if len(contents) > 10:
            print(f"      ... and {len(contents) - 10} more items")
    except Exception as e:
        print(f"   ❌ Error reading competitionData: {e}")

# Check output directory
print(f"\n5. Output Directory:")
output_path = baseDir + '/derived/tfRecords'
exists = os.path.exists(output_path)
status = "✅ EXISTS" if exists else "⚠️  WILL BE CREATED"
print(f"   {status} - derived/tfRecords/")

print("\n" + "=" * 60)
print("SUMMARY")
print("=" * 60)

# Create summary
all_good = True
issues = []

if not os.path.exists(baseDir):
    all_good = False
    issues.append("❌ Base directory not found")

if not os.path.exists(baseDir + '/makeTFRecordsFromSession.py'):
    all_good = False
    issues.append("❌ makeTFRecordsFromSession.py missing")

if not os.path.exists(baseDir + '/getSpeechSessionBlocks.py'):
    all_good = False
    issues.append("❌ getSpeechSessionBlocks.py missing")

if not os.path.exists(neural_decoder_path):
    all_good = False
    issues.append("❌ neuralDecoder folder missing")

if not os.path.exists(competition_data_path):
    all_good = False
    issues.append("❌ competitionData folder missing")

if all_good:
    print("✅ ALL REQUIRED FILES AND FOLDERS FOUND!")
    print("✅ Ready to run Cell 3!")
else:
    print("⚠️  ISSUES FOUND:")
    for issue in issues:
        print(f"   {issue}")
    print("\n📝 You need to upload the missing files/folders to Google Drive")

print("=" * 60)
```

```python
# Import your custom modules
from makeTFRecordsFromSession import makeTFRecordsFromCompetitionFiles
from getSpeechSessionBlocks import getSpeechSessionBlocks

# Get block lists and process
blockLists = getSpeechSessionBlocks()

for sessIdx in range(len(blockLists)):
    sessionName = blockLists[sessIdx][0]
    dataPath = baseDir + '/competitionData'
    tfRecordFolder = baseDir + '/derived/tfRecords/' + sessionName

    print(f"Processing session {sessIdx + 1}/{len(blockLists)}: {sessionName}")
    try:
        makeTFRecordsFromCompetitionFiles(sessionName, dataPath, tfRecordFolder)
        print(f"✓ Completed: {sessionName}")
    except Exception as e:
        print(f"✗ Error on {sessionName}: {e}")
        # Continue with next session instead of crashing
        continue

print("All sessions processed!")
```

```python
# Check what was actually created in derived/tfRecords
import os

print("=" * 70)
print("CHECKING TFRECORDS OUTPUT")
print("=" * 70)

tfrecords_path = baseDir + '/derived/tfRecords'

print(f"\n1. Does derived/tfRecords exist?")
print(f"   Path: {tfrecords_path}")
print(f"   Exists: {os.path.exists(tfrecords_path)}")

if os.path.exists(tfrecords_path):
    print(f"\n2. Contents of derived/tfRecords:")
    contents = os.listdir(tfrecords_path)
    print(f"   Found {len(contents)} items")

    if len(contents) == 0:
        print("   ⚠️  FOLDER IS EMPTY!")
        print("\n   This means the processing completed but didn't create any files.")
        print("   Check Cell 4 output for error messages.")
    else:
        print(f"\n   Items found:")
        for item in sorted(contents):
            item_path = os.path.join(tfrecords_path, item)
            if os.path.isdir(item_path):
                # Count files inside each session folder
                try:
                    files = os.listdir(item_path)
                    print(f"   📁 {item}/ - ({len(files)} files)")
                except:
                    print(f"   📁 {item}/ - (error reading)")
            else:
                print(f"   📄 {item}")

        # Show detailed contents of first session
        if len(contents) > 0 and os.path.isdir(os.path.join(tfrecords_path, contents[0])):
            first_session = contents[0]
            first_session_path = os.path.join(tfrecords_path, first_session)
            print(f"\n3. Detailed view of {first_session}/:")
            for file in os.listdir(first_session_path):
                file_path = os.path.join(first_session_path, file)
                file_size = os.path.getsize(file_path)
                print(f"      - {file} ({file_size} bytes)")
else:
    print("   ❌ FOLDER DOESN'T EXIST!")
    print("   The folder wasn't even created. Check if Cell 2 ran successfully.")

print("\n" + "=" * 70)
```
